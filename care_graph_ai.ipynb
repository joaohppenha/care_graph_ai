##Executive Summary & Problem Statement

Large Language Models (LLMs) and environmental RAG systems trained on ecological datasets frequently leak sensitive geospatial coordinates of vulnerable indigenous territories (exposing isolated communities to illegal mining, poaching, and land invasions) while enabling unauthorized commercial appropriation of traditional ecological knowledge (**biopiracy**).

Current generative AI frameworks lack deterministic, enforceable safeguards aligned with indigenous data governance rights and national biodiversity protection laws.

### 1. **Formal OWL ontology**
Initializes and exports the formal OWL ontology using owlready2 by defining core semantic classes such as ConhecimentoEcologico, Territorio, PovoOriginario, NivelVulnerabilidade, and PoliticaDeAcesso, establishing object properties for relational mapping like protegidoPor, pertenceA, habitadoPor, and temVulnerabilidade, instantiating real-world entities such as the Javari Valley and Xingu Indigenous Park with critical and moderate vulnerability levels alongside traditional knowledge assets, and finally serializing the complete ontological graph into the care_sovereignty.owl XML format file.

```python
# ----------------------------------------------------
# 1. ONTOLOGY INITIALIZATION (owlready2)
# ----------------------------------------------------
print("Initializing ontological graph...")
onto_path.append(".")
onto = get_ontology("http://test.org/care_sovereignty.owl")

with onto:
    class ConhecimentoEcologico(Thing): pass
    class Territorio(Thing): pass
    class PovoOriginario(Thing): pass
    class NivelVulnerabilidade(Thing): pass
    class PoliticaDeAcesso(Thing): pass

    class protegidoPor(ConhecimentoEcologico >> PoliticaDeAcesso): pass
    class pertenceA(ConhecimentoEcologico >> Territorio): pass
    class habitadoPor(Territorio >> PovoOriginario): pass
    class temVulnerabilidade(Territorio >> NivelVulnerabilidade): pass

    vuln_critica = NivelVulnerabilidade("VulnerabilidadeCritica")
    vuln_moderada = NivelVulnerabilidade("VulnerabilidadeModerada")

    javari = Territorio("TerritorioIndigenaValeDoJavari")
    javari.temVulnerabilidade = [vuln_critica]

    xingu = Territorio("ParqueIndigenaDoXingu")
    xingu.temVulnerabilidade = [vuln_moderada]

    mao_diretor = PovoOriginario("PovosMatisEValeriaIsolados")
    javari.habitadoPor = [mao_diretor]

    povo_xinguano = PovoOriginario("PovosXinguanos")
    xingu.habitadoPor = [povo_xinguano]

    politica_restrita = PoliticaDeAcesso("PoliticaRestricaoComercialEBiopirataria")
    politica_pesquisa = PoliticaDeAcesso("PoliticaPesquisaAcademicaOfuscada")
    politica_oficial = PoliticaDeAcesso("PoliticaOrgaoOficialControlado")

    sab_antropica = ConhecimentoEcologico("ManejoTerraPretaAntropica")
    sab_antropica.pertenceA = [javari]
    sab_antropica.protegidoPor = [politica_restrita]

    sab_fenologico = ConhecimentoEcologico("IndicadoresFenologicosDeCheias")
    sab_fenologico.pertenceA = [xingu]
    sab_fenologico.protegidoPor = [politica_pesquisa]

onto.save(file="care_sovereignty.owl", format="rdfxml")
print("Ontology saved successfully as 'care_sovereignty.owl'.")
```

###**2. Deterministic Governance Engine**

Defining the Pydantic structured output contract (`GovernanceContract`), the real-time immutable audit logger (`audit_logger`) that writes to `governance_audit.jsonl`, and the core validation logic (`evaluate_governance_engine`) that enforces strict rules against commercial biopiracy without FPIC, applies geospatial obfuscation for academic research in vulnerable indigenous territories, and authorizes full regulatory access for official oversight agencies.

```python
# ----------------------------------------------------
# 2. GOVERNANCE ENGINE & PYDANTIC CONTRACTS (Fixed Datetime)
# ----------------------------------------------------
from datetime import datetime, timezone

class GovernanceContract(BaseModel):
    status: Literal["ACCESS_DENIED", "ACCESS_GRANTED_OBFUSCATED", "ACCESS_GRANTED_FULL"] = Field(
        ..., description="Deterministic firewall decision based on ontology rules and biodiversity legislation."
    )
    reason: str = Field(..., description="Legal and ethical justification referencing CARE Principles and Brazilian Law 13.123/2015.")
    safe_coordinates: str = Field(..., description="Obscured watershed boundaries or precise geographic coordinates based on access tier.")
    response_content: str = Field(..., description="Secure output payload adhering to access restrictions.")

def audit_logger(log_data: dict):
    os.makedirs("logs", exist_ok=True)
    with open("logs/governance_audit.jsonl", "a", encoding="utf-8") as f:
        f.write(json.dumps(log_data, ensure_ascii=False) + "\n")

def evaluate_governance_engine(intent: str, requester_type: str, territory: str) -> GovernanceContract:
    timestamp = datetime.now(timezone.utc).isoformat()

    if intent.lower() == "commercial" or requester_type.lower() == "commercial enterprise":
        contract = GovernanceContract(
            status="ACCESS_DENIED",
            reason="Blocked by CARE-Graph Firewall: Commercial use without Free, Prior and Informed Consent (FPIC) and SisGen registration violates Brazilian Biodiversity Law (Law 13.123/2015) and Indigenous Data Sovereignty principles.",
            safe_coordinates="[BLOCKED - SENSITIVE GEOSPATIAL DATA PROTECTED]",
            response_content="Access Denied. Traditional knowledge and precise indigenous geographic coordinates cannot be exploited for commercial purposes without formal benefit-sharing agreements and community consent."
        )
    elif "javari" in territory.lower() and requester_type.lower() == "academic researcher":
        contract = GovernanceContract(
            status="ACCESS_GRANTED_OBFUSCATED",
            reason="Approved with spatial obfuscation: Academic research allowed, but exact coordinates suppressed to protect isolated indigenous communities from physical exposure.",
            safe_coordinates="Amazon Basin / Javari Watershed Region (Precise GPS coordinates suppressed per Critical Vulnerability protocol).",
            response_content="Traditional ecological knowledge regarding anthropic dark earth management has been successfully retrieved. Precise geographic vectors have been replaced with macro-watershed boundaries to safeguard territorial sovereignty."
        )
    elif requester_type.lower() == "official agency (funai/ibama)":
        contract = GovernanceContract(
            status="ACCESS_GRANTED_FULL",
            reason="Full access authorized: Regulatory compliance verification by authorized federal agency under official mandate.",
            safe_coordinates="Exact Coordinates: -4.3129° S, 70.0983° W (Vale do Javari Border Sector)",
            response_content="[Authorized Audit View] Full environmental monitoring data and exact territorial boundaries provided for official state oversight, compliance tracking, and defensive demarcation."
        )
    else:
        contract = GovernanceContract(
            status="ACCESS_GRANTED_OBFUSCATED",
            reason="Default defensive posture applied.",
            safe_coordinates="General Regional Watershed Indicator",
            response_content="Information provided under standard sovereign data sharing rules."
        )

    audit_record = {
        "timestamp": timestamp,
        "intent": intent,
        "requester": requester_type,
        "territory": territory,
        "decision": contract.status,
        "reason": contract.reason
    }
    audit_logger(audit_record)
    return contract
```

### **3. Interactive visual graph using NetworkX and PyVis**

instantiating semantic nodes for vulnerable indigenous territories, traditional knowledge assets, and legal frameworks, linking them with directed relational edges, and exporting the rendered simulation into an interactive HTML file (`ontology_graph.html`).

```python
!pip install -q folium
```

```python
# ----------------------------------------------------
# 3. INTERACTIVE VISUAL GRAPH (NetworkX + Matplotlib)
# ----------------------------------------------------
import matplotlib.pyplot as plt

def generate_static_graph():
    plt.figure(figsize=(10, 5))
    G = nx.DiGraph()

    # Adicionando nós
    nodes = {
        "Vale do Javari\n(Critical Vulnerability)": "#ff4d4d",
        "Parque Xingu\n(Moderate Vulnerability)": "#ffcc00",
        "Manejo Terra Preta\n(Protected Knowledge)": "#4da6ff",
        "Indicadores Fenológicos\n(Academic Access)": "#4da6ff",
        "Lei 13.123/2015\n(Biopiracy Firewall)": "#00cc66",
        "CARE Principles\n(Indigenous Sovereignty)": "#00cc66"
    }

    for node in nodes:
        G.add_node(node)

    # Arestas
    G.add_edge("Manejo Terra Preta\n(Protected Knowledge)", "Vale do Javari\n(Critical Vulnerability)")
    G.add_edge("Manejo Terra Preta\n(Protected Knowledge)", "Lei 13.123/2015\n(Biopiracy Firewall)")
    G.add_edge("Indicadores Fenológicos\n(Academic Access)", "Parque Xingu\n(Moderate Vulnerability)")
    G.add_edge("Vale do Javari\n(Critical Vulnerability)", "CARE Principles\n(Indigenous Sovereignty)")
    G.add_edge("Parque Xingu\n(Moderate Vulnerability)", "CARE Principles\n(Indigenous Sovereignty)")

    pos = nx.spring_layout(G, seed=42)
    colors = [nodes[node] for node in G.nodes()]

    nx.draw(G, pos, with_labels=True, node_color=colors, node_size=3000, font_size=9, font_weight="bold", font_color="white", edge_color="#888888", width=2)
    plt.title("CARE-Graph AI: Ontological Governance & Territorial Protection Map", color="white", fontsize=12)
    plt.gca().set_facecolor("#1a1a1a")
    plt.gcf().patch.set_facecolor("#1a1a1a")

    path = "ontology_graph.png"
    plt.savefig(path, bbox_inches="tight", facecolor=plt.gcf().get_facecolor(), edgecolor='none')
    plt.close()
    return path

    # ----------------------------------------------------
# 3.1. GEOSPATIAL MAP GENERATION (Folium)
# ----------------------------------------------------
import folium
import os

def generate_satellite_map(status: str):
    lat, lon = -4.3129, -70.0983

    # Define zoom e estilo baseados na decisão de governança
    if status == "ACCESS_DENIED":
        zoom = 4
        color = "#ff4d4d"
        popup_text = "<b>Vale do Javari</b><br>STATUS: ACCESS_DENIED<br><i>Coordenadas protegidas contra exploração comercial.</i>"
    elif "OBFUSCATED" in status:
        zoom = 6
        color = "#ffcc00"
        popup_text = "<b>Vale do Javari</b><br>STATUS: ACCESS_GRANTED_OBFUSCATED<br><i>Coordenadas de precisão ofuscadas (Proteção de povos isolados).</i>"
    else:
        zoom = 11
        color = "#00cc66"
        popup_text = f"<b>Vale do Javari</b><br>STATUS: ACCESS_GRANTED_FULL<br><i>Auditoria Oficial - Coordenadas: {lat}, {lon}</i>"

    # Cria o mapa centrado na região com camada de satélite real da Esri
    m = folium.Map(
        location=[lat, lon],
        zoom_start=zoom,
        tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}',
        attr='Esri World Imagery'
    )

    # Adiciona o marcador interativo
    folium.CircleMarker(
        location=[lat, lon],
        radius=20,
        popup=popup_text,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.85
    ).add_to(m)

    os.makedirs("maps", exist_ok=True)
    map_path = "maps/satellite_map.html"
    m.save(map_path)

    # Retorna o conteúdo HTML pronto para ser embutido com segurança no Gradio
    with open(map_path, "r", encoding="utf-8") as f:
        html_content = f.read()

    # Encapsula em um iframe srcdoc para evitar qualquer conflito de script no front-end do Gradio
    safe_html = f"<iframe srcdoc='{html_content.replace(chr(39), chr(34))}' width='100%' height='420px' style='border:none; border-radius:8px;'></iframe>"
    return safe_html
```

### **4. Gradio web interface**
Featuring two main tabs: the first tab renders the interactive PyVis knowledge graph alongside architectural documentation, and the second tab provides a one-click simulation dashboard with three test scenarios that trigger the deterministic governance engine, evaluate access requests, update geospatial obfuscation rules, and display real-time JSONL audit logs.

```python

# ----------------------------------------------------
# 4. GRADIO INTERFACE SIMULATION LOGIC
# ----------------------------------------------------
def run_simulation_with_satellite(simulation_mode):
    try:
        if simulation_mode == "Commercial Biopiracy Attempt":
            res = evaluate_governance_engine(intent="commercial", requester_type="Commercial Enterprise", territory="Vale do Javari")
        elif simulation_mode == "Academic Climate Research":
            res = evaluate_governance_engine(intent="research", requester_type="Academic Researcher", territory="Vale do Javari")
        else:
            res = evaluate_governance_engine(intent="audit", requester_type="Official Agency (FUNAI/IBAMA)", territory="Vale do Javari")

        logs_output = ""
        log_path = "logs/governance_audit.jsonl"
        if os.path.exists(log_path):
            with open(log_path, "r", encoding="utf-8") as f:
                lines = f.readlines()
                logs_output = "".join(lines[-5:]) if lines else "No audit logs recorded yet."

        bg_color = '#ff4d4d' if res.status == 'ACCESS_DENIED' else '#ffcc00' if 'OBFUSCATED' in res.status else '#00cc66'
        status_html = f"<div style='padding: 12px; border-radius: 6px; background-color: {bg_color}; color: white; font-weight: bold; text-align: center;'>STATUS: {res.status}</div>"

        satellite_widget = generate_satellite_map(res.status)

        return status_html, res.reason, res.safe_coordinates, res.response_content, logs_output, satellite_widget
    except Exception as e:
        err_msg = str(e)
        return f"<div style='padding: 12px; background-color: red; color: white;'>ERROR: {err_msg}</div>", err_msg, err_msg, err_msg, err_msg, "<p>Erro ao carregar mapa.</p>"

# ----------------------------------------------------
# 5. GRADIO BLOCKS APP (Complete, English & No Emojis)
# ----------------------------------------------------
with gr.Blocks() as demo:
    gr.Markdown("# CARE-Graph AI")
    gr.Markdown("### Neuro-Symbolic Firewall for Indigenous Data Sovereignty & Territorial Protection")
    gr.Markdown("Submitted to Code for Humanity Hackathon | Combining OWL Ontologies, Pydantic Governance Contracts, and Secure Local Execution.")

    with gr.Tabs():
        with gr.TabItem("Architecture & Knowledge Graph"):
            gr.Markdown("#### Ontological Governance Map")
            gr.Markdown("Visual representation of traditional knowledge nodes linked to protected indigenous territories and legal firewalls.")
            graph_img_path = generate_static_graph()
            gr.Image(value=graph_img_path, label="Ontology & Governance Network")

            gr.Markdown("""
            ### Ontological Governance Map: Architecture & Technical Rationale

            This directed graph illustrates the neuro-symbolic integration between the **OWL Ontology (`owlready2`)**, **Brazilian Biodiversity Legislation (Law 13.123/2015)**, and the **CARE Principles** for Indigenous Data Sovereignty:

            * **Blue Nodes (Protected Knowledge & Academic Access):**
              * *Justification:* Represent traditional ecological knowledge assets (e.g., Anthropic Dark Earth Management and Phenological Indicators) extracted from the knowledge base. They serve as informational assets under strict custodial governance rather than public domain commodities.
            * **Red & Yellow Nodes (Critical & Moderate Vulnerability):**
              * *Justification:* Map high-sensitivity indigenous territories (such as Vale do Javari and Xingu Park). These nodes dictate the geographic restriction tier and trigger automated spatial obfuscation logic to prevent the physical exposure of isolated communities.
            * **Green Nodes (Firewalls & Sovereignty):**
              * *Justification:* Enforce definitive governance contracts rooted in Law 13.123/2015 (combating biopiracy) and the CARE Principles (Collective benefit, Authority to control, Responsibility, Ethics). They act as a symbolic barrier that blocks unauthorized commercial exploitation while permitting verified academic or state oversight workflows.
            """)

        with gr.TabItem("Live Governance Firewall & Satellite Intelligence"):
            gr.Markdown("#### 1-Click Demo Simulator with Satellite Geospatial Enforcement")
            gr.Markdown("Select a scenario to trigger the firewall and update real satellite imagery permissions in real time:")

            with gr.Row():
                btn1 = gr.Button("Commercial Biopiracy Attempt", variant="stop")
                btn2 = gr.Button("Academic Climate Research", variant="primary")
                btn3 = gr.Button("Official Agency — FUNAI/IBAMA", variant="secondary")

            with gr.Row():
                with gr.Column():
                    out_status = gr.HTML(label="Firewall Status")
                    out_reason = gr.Textbox(label="Ontological & Legal Justification", lines=3)
                    out_coords = gr.Textbox(label="Geospatial Governance (Coordinates / Obfuscation)", lines=2)
                    out_map = gr.HTML(label="Sovereign Satellite Intelligence Map")
                with gr.Column():
                    out_llm = gr.Textbox(label="Controlled Secure Output", lines=4)
                    out_logs = gr.Code(label="Live JSONL Audit Trail (governance_audit.jsonl)", language="json")

            outputs_list = [out_status, out_reason, out_coords, out_llm, out_logs, out_map]

            btn1.click(fn=lambda: run_simulation_with_satellite("Commercial Biopiracy Attempt"), outputs=outputs_list)
            btn2.click(fn=lambda: run_simulation_with_satellite("Academic Climate Research"), outputs=outputs_list)
            btn3.click(fn=lambda: run_simulation_with_satellite("Official Agency — FUNAI/IBAMA"), outputs=outputs_list)

if __name__ == "__main__":
    demo.launch(inline=False, share=True, theme=gr.themes.Soft())
```
